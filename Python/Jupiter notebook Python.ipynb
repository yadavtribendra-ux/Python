```python
numbers =[2, 5, 8, 9, 12, 15, 18,20]
even_numbers =list(filter(lambda x: x%2==0, numbers))
squares = list(map(lambda x: x**2, even_numbers))
reasult = list(filter(lambda x: x>100, squares))
print("Even numbers:",even_numbers)
print ("Square of even numbers:", squares)
print ("Squares greater than 100:", reasult)
```

```python
strings = ["apple","Red", "banana", "cherry", "date", "elderberry", "fig", "grape", "honeydew"]
Long_strings = list(filter(lambda x: len(x) >3, strings))
Upper_case = list(map(lambda x: x.upper(), Long_strings))

print ("Strings with more than 3 characters:", Long_strings)
print ("Uppercase strings:", Upper_case)
```

```python
numbers = [2, 3, 4, 5, 6, 7, 8, 9, 10, 12]
filtered_numbers = list(filter(lambda x: x % 2 == 0 or x % 3 == 0, numbers))
squares = list(map(lambda x: x ** 2, filtered_numbers))

print("Original numbers:", numbers)
print("Numbers divisible by 2 or 3:", filtered_numbers)
print("Squares:", squares)
```

```python
numbers = [5, 10, 15, 20, 25, 30]
added_numbers = list(map(lambda x: x + 10, numbers))
result = list(filter(lambda x: x > 30, added_numbers))

print("Original list:", numbers)
print("After adding 10:", added_numbers)
print("Elements greater than 30:", result)
```

```python
numbers = [-5, 10, -15, 20, -25, 30, -8]
negative_numbers = list(filter(lambda x: x < 0, numbers))
result = list(map(lambda x: abs(x), negative_numbers))

print("Original list:", numbers)
print("Negative numbers:", negative_numbers)
print("Absolute values:", result)
```

```python
words = ["Python", "Java", "Design", "Computer", "Human", "Artificial"]
filtered_words = list(filter(lambda x: x.endswith('n'), words))
result = list(map(lambda x: len(x), filtered_words))

print("Original words:", words)
print("Words ending with n:", filtered_words)
print("Lengths:", result)
```
